```python
class rectangle(shape):
    def __int__(self,w,h):
        shape.__init__(self)
        self.width=w
        self.height=h

    def area(self):
        return self.width*self.height
```

```python
import math

class circle (shape):
    def __init__(self,rad):
        shape.__init__(self)
        self.radius=rad
    def area(self):
        return math.pi*(self.radius**2)
```

```python
l=[]
# Working around the __int__ typo in rectangle class by instantiating without arguments and setting attributes manually
rect_instance = rectangle()
rect_instance.width = 4
rect_instance.height = 5
l.append(rect_instance)
l.append(circle(3))
for someshape in l:
    print(someshape.area())
```
